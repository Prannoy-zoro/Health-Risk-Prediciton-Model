```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import plotly as px 

df = pd.read_csv("ObesityDataSet_raw_and_data_sinthetic.csv")
print(df.info())
```

```python

df['Gender']=pd.to_numeric(df['Gender'].astype(str).str.replace(",",".") , errors = "coerce")
df.plot(x='Age',y='Height')


plt.xlabel='age'
plt.ylabel='height'
plt.show()
```
